# 02 — Exploratory Data Analysis

Explore distributions, relationships, and patterns in the lifestyle and well-being variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/mental_wellbeing.csv')
df.head()
```

```python
df.describe(include='all').T
```

```python
# Numeric correlation matrix
numeric_df = df.select_dtypes(include='number')
if not numeric_df.empty:
    plt.figure(figsize=(10, 7))
    sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f')
    plt.title('Correlation Matrix')
    plt.show()
```

## Questions to investigate
- Which lifestyle variables are most strongly related to the target?
- Are there outliers?
- Are numerical variables skewed?
- Do categorical groups show meaningful differences in the target?
