# 01 — Data Loading & Cleaning

Load the mental well-being dataset, inspect its structure, identify missing/duplicate values, and prepare a clean dataset.

```python
import pandas as pd
import numpy as np

DATA_PATH = "../../data/mental_wellbeing.csv"

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
import os

print("Current folder:")
print(os.getcwd())

print("\nFile exists?")
print(os.path.exists("data/mental_wellbeing.csv"))
```

```python
import os

print(os.listdir("../../data"))
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
print("All columns with their data types:\n")

for i, col in enumerate(df.columns, start=1):
    print(i, "->", col, "->", df[col].dtype)
```

```python
print("Unique values in categorical columns:\n")

print("DAILY_STRESS:")
print(df["DAILY_STRESS"].unique())

print("\nAGE:")
print(df["AGE"].unique())

print("\nGENDER:")
print(df["GENDER"].unique())

print("\nTarget statistics:")
print(df["WORK_LIFE_BALANCE_SCORE"].describe())
```

```python
print("DAILY_STRESS value counts:")
print(df["DAILY_STRESS"].value_counts())

print("\nAGE value counts:")
print(df["AGE"].value_counts())

print("\nGENDER value counts:")
print(df["GENDER"].value_counts())

print("\nNumerical feature ranges:")
print(df.select_dtypes(include="number").describe().T)
```

```python
print("Row with invalid DAILY_STRESS value:")

print(df[df["DAILY_STRESS"] == "1/1/00"])
```

```python
# Remove the row with invalid DAILY_STRESS value
df = df[df["DAILY_STRESS"] != "1/1/00"].copy()

print("Shape after cleaning:", df.shape)
print("Invalid DAILY_STRESS values remaining:",
      (df["DAILY_STRESS"] == "1/1/00").sum())
```
